## print()

**Descripción y uso:** Muestra información en la consola. Es útil para comunicar resultados y verificar valores durante la ejecución.

**Sintaxis:**
```python
print(*objetos, sep=' ', end='\n')
```

**Parámetros:** objetos: uno o varios valores que se desean mostrar; sep: separador; end: texto final.

**Ejemplo práctico:**

````python
nombre = "Laura"
edad = 21
print("Nombre:", nombre)
print("Edad:", edad)
````

**Resultado esperado:**
```text
Nombre: Laura
Edad: 21
```

## input()

**Descripción y uso:** Solicita al usuario que escriba un dato y devuelve lo ingresado como una cadena de texto (str).

**Sintaxis:**
```python
input(prompt)
```

**Parámetros:** prompt: mensaje opcional que aparece antes de recibir el dato.

**Ejemplo práctico:**

````python
nombre = input("¿Cómo te llamas? ")
print("Hola,", nombre)
````

**Resultado esperado:**
```text
Si el usuario escribe Ana:
Hola, Ana
```

## type()

**Descripción y uso:** Indica el tipo de dato de un objeto. Ayuda a identificar si un valor es entero, texto, lista, etc.

**Sintaxis:**
```python
type(objeto)
```

**Parámetros:** objeto: valor o variable cuyo tipo se desea conocer.

**Ejemplo práctico:**

````python
numero = 25
texto = "Python"
print(type(numero))
print(type(texto))
````

**Resultado esperado:**
```text
<class 'int'>
<class 'str'>
```

## int()

**Descripción y uso:** Convierte un valor compatible a un número entero. En números decimales elimina la parte fraccionaria, no redondea.

**Sintaxis:**
```python
int(valor)
```

**Parámetros:** valor: número o cadena numérica que se desea convertir.

**Ejemplo práctico:**

````python
edad = int("19")
precio = int(8.9)
print(edad)
print(precio)
````

**Resultado esperado:**
```text
19
8
```

## float()

**Descripción y uso:** Convierte un valor compatible a un número de punto flotante, útil para representar decimales.

**Sintaxis:**
```python
float(valor)
```

**Parámetros:** valor: número o cadena numérica convertible.

**Ejemplo práctico:**

````python
altura = float("1.72")
resultado = float(5)
print(altura)
print(resultado)
````

**Resultado esperado:**
```text
1.72
5.0
```

## str()

**Descripción y uso:** Convierte un objeto a su representación en texto. Es útil para combinar valores con mensajes.

**Sintaxis:**
```python
str(objeto)
```

**Parámetros:** objeto: valor que se desea representar como cadena.

**Ejemplo práctico:**

````python
edad = 20
mensaje = "Tengo " + str(edad) + " años"
print(mensaje)
````

**Resultado esperado:**
```text
Tengo 20 años
```

## bool()

**Descripción y uso:** Convierte un valor a verdadero (True) o falso (False) según las reglas de veracidad de Python.

**Sintaxis:**
```python
bool(valor)
```

**Parámetros:** valor: objeto que se evaluará como verdadero o falso.

**Ejemplo práctico:**

````python
print(bool(0))
print(bool(15))
print(bool(""))
print(bool("Python"))
````

**Resultado esperado:**
```text
False
True
False
True
```

## len()

**Descripción y uso:** Devuelve la cantidad de elementos de una secuencia o colección, como una cadena, lista o diccionario.

**Sintaxis:**
```python
len(objeto)
```

**Parámetros:** objeto: secuencia o colección cuya longitud se quiere conocer.

**Ejemplo práctico:**

````python
fruta = "manzana"
colores = ["rojo", "azul", "verde"]
print(len(fruta))
print(len(colores))
````

**Resultado esperado:**
```text
7
3
```

## sum()

**Descripción y uso:** Suma los elementos de un iterable numérico. Puede recibir un valor inicial opcional.

**Sintaxis:**
```python
sum(iterable, start=0)
```

**Parámetros:** iterable: colección de números; start: valor desde el que comienza la suma.

**Ejemplo práctico:**

````python
calificaciones = [4.0, 3.5, 4.5]
total = sum(calificaciones)
print("Total:", total)
print("Con valor inicial:", sum([2, 3], 10))
````

**Resultado esperado:**
```text
Total: 12.0
Con valor inicial: 15
```

## max()

**Descripción y uso:** Devuelve el elemento mayor de un iterable o el mayor entre varios argumentos.

**Sintaxis:**
```python
max(iterable)  |  max(valor1, valor2, ...)
```

**Parámetros:** iterable o varios valores comparables. También admite key y default en ciertos usos.

**Ejemplo práctico:**

````python
temperaturas = [18, 25, 21, 29]
print(max(temperaturas))
print(max(7, 12, 4))
````

**Resultado esperado:**
```text
29
12
```
