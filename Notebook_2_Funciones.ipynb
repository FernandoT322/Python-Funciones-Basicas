## min()

**Descripción y uso:** Devuelve el elemento menor de un iterable o el menor entre varios argumentos.

**Sintaxis:**
```python
min(iterable)  |  min(valor1, valor2, ...)
```

**Parámetros:** iterable o varios valores comparables.

**Ejemplo práctico:**

````python
precios = [12000, 9500, 15000]
print(min(precios))
print(min(8, 3, 6))
````

**Resultado esperado:**
```text
9500
3
```

## abs()

**Descripción y uso:** Devuelve el valor absoluto de un número, es decir, su distancia respecto de cero sin considerar el signo.

**Sintaxis:**
```python
abs(numero)
```

**Parámetros:** numero: valor numérico.

**Ejemplo práctico:**

````python
diferencia = -17
print(abs(diferencia))
print(abs(-3.5))
````

**Resultado esperado:**
```text
17
3.5
```

## round()

**Descripción y uso:** Redondea un número al entero más cercano o a una cantidad indicada de cifras decimales.

**Sintaxis:**
```python
round(numero, ndigits=None)
```

**Parámetros:** numero: valor que se redondea; ndigits: cantidad de decimales (opcional).

**Ejemplo práctico:**

````python
precio = 18.7564
print(round(precio))
print(round(precio, 2))
````

**Resultado esperado:**
```text
19
18.76
```

## pow()

**Descripción y uso:** Calcula una potencia. Con tres argumentos, calcula la potencia y aplica el módulo.

**Sintaxis:**
```python
pow(base, exponente[, modulo])
```

**Parámetros:** base: número base; exponente: potencia; modulo: divisor opcional para el residuo.

**Ejemplo práctico:**

````python
print(pow(2, 3))
print(pow(2, 5, 3))
````

**Resultado esperado:**
```text
8
2
```

## range()

**Descripción y uso:** Genera una secuencia inmutable de enteros, comúnmente usada para repetir instrucciones en un ciclo for.

**Sintaxis:**
```python
range(stop)  |  range(start, stop[, step])
```

**Parámetros:** start: inicio (opcional); stop: límite excluido; step: incremento (opcional).

**Ejemplo práctico:**

````python
for numero in range(1, 6):
    print(numero)
````

**Resultado esperado:**
```text
1
2
3
4
5
```

## sorted()

**Descripción y uso:** Devuelve una nueva lista con los elementos de un iterable ordenados, sin modificar el original.

**Sintaxis:**
```python
sorted(iterable, key=None, reverse=False)
```

**Parámetros:** iterable: colección; key: criterio de ordenamiento opcional; reverse: orden descendente si es True.

**Ejemplo práctico:**

````python
notas = [3.8, 4.5, 2.9, 4.0]
ordenadas = sorted(notas)
print(ordenadas)
print(sorted(notas, reverse=True))
````

**Resultado esperado:**
```text
[2.9, 3.8, 4.0, 4.5]
[4.5, 4.0, 3.8, 2.9]
```

## list()

**Descripción y uso:** Crea una lista. Si recibe un iterable, construye una lista con sus elementos.

**Sintaxis:**
```python
list(iterable=())
```

**Parámetros:** iterable: objeto iterable opcional, como una tupla o cadena.

**Ejemplo práctico:**

````python
vocales = list("aeiou")
print(vocales)
print(list((10, 20, 30)))
````

**Resultado esperado:**
```text
['a', 'e', 'i', 'o', 'u']
[10, 20, 30]
```

## tuple()

**Descripción y uso:** Crea una tupla, una colección ordenada que no puede modificarse después de crearla.

**Sintaxis:**
```python
tuple(iterable=())
```

**Parámetros:** iterable: objeto iterable opcional.

**Ejemplo práctico:**

````python
coordenadas = tuple([4, 8])
print(coordenadas)
print(tuple("sol"))
````

**Resultado esperado:**
```text
(4, 8)
('s', 'o', 'l')
```

## set()

**Descripción y uso:** Crea un conjunto, colección sin elementos repetidos y sin orden posicional garantizado.

**Sintaxis:**
```python
set(iterable=())
```

**Parámetros:** iterable: colección opcional cuyos elementos sean hashables.

**Ejemplo práctico:**

````python
numeros = set([1, 2, 2, 3, 3, 3])
print(sorted(numeros))
print(len(numeros))
````

**Resultado esperado:**
```text
[1, 2, 3]
3
```

## dict()

**Descripción y uso:** Crea un diccionario de pares clave-valor. Las claves deben ser únicas y hashables.

**Sintaxis:**
```python
dict(**kwargs)  |  dict(iterable)
```

**Parámetros:** Puede recibir argumentos con nombre, pares clave-valor o un diccionario existente.

**Ejemplo práctico:**

````python
estudiante = dict(nombre="Sofía", edad=20)
print(estudiante)
print(estudiante["nombre"])
````

**Resultado esperado:**
```text
{'nombre': 'Sofía', 'edad': 20}
Sofía
```
