## enumerate()

**Descripción y uso:** Recorre un iterable y entrega parejas con el índice y el elemento correspondiente.

**Sintaxis:**
```python
enumerate(iterable, start=0)
```

**Parámetros:** iterable: colección que se recorre; start: índice inicial opcional.

**Ejemplo práctico:**

````python
materias = ["Matemáticas", "Historia", "Programación"]
for indice, materia in enumerate(materias, start=1):
    print(indice, materia)
````

**Resultado esperado:**
```text
1 Matemáticas
2 Historia
3 Programación
```

## zip()

**Descripción y uso:** Agrupa elementos de varios iterables posición por posición. Termina cuando se agota el iterable más corto.

**Sintaxis:**
```python
zip(*iterables, strict=False)
```

**Parámetros:** iterables: dos o más colecciones; strict: si es True, genera error si las longitudes difieren.

**Ejemplo práctico:**

````python
nombres = ["Luis", "Marta", "Juan"]
notas = [4.2, 4.8, 3.9]
for nombre, nota in zip(nombres, notas):
    print(nombre, nota)
````

**Resultado esperado:**
```text
Luis 4.2
Marta 4.8
Juan 3.9
```

## map()

**Descripción y uso:** Aplica una función a cada elemento de uno o más iterables y devuelve un iterador con los resultados.

**Sintaxis:**
```python
map(funcion, iterable, ...)
```

**Parámetros:** funcion: operación que se aplicará; iterable: colección o colecciones de entrada.

**Ejemplo práctico:**

````python
numeros = [1, 2, 3, 4]
dobles = map(lambda n: n * 2, numeros)
print(list(dobles))
````

**Resultado esperado:**
```text
[2, 4, 6, 8]
```

## filter()

**Descripción y uso:** Selecciona los elementos de un iterable para los que una función de prueba devuelve True.

**Sintaxis:**
```python
filter(funcion, iterable)
```

**Parámetros:** funcion: predicado que devuelve True o False; iterable: colección que se filtra.

**Ejemplo práctico:**

````python
edades = [12, 18, 21, 15, 30]
mayores = filter(lambda edad: edad >= 18, edades)
print(list(mayores))
````

**Resultado esperado:**
```text
[18, 21, 30]
```

## any()

**Descripción y uso:** Devuelve True si al menos un elemento de un iterable es verdadero. Devuelve False si ninguno lo es.

**Sintaxis:**
```python
any(iterable)
```

**Parámetros:** iterable: colección de valores o condiciones.

**Ejemplo práctico:**

````python
respuestas = [False, False, True]
print(any(respuestas))
print(any([0, "", None]))
````

**Resultado esperado:**
```text
True
False
```

## all()

**Descripción y uso:** Devuelve True si todos los elementos de un iterable son verdaderos. Si encuentra uno falso, devuelve False.

**Sintaxis:**
```python
all(iterable)
```

**Parámetros:** iterable: colección de valores o condiciones.

**Ejemplo práctico:**

````python
edades = [20, 22, 19]
print(all(edad >= 18 for edad in edades))
print(all([True, True, False]))
````

**Resultado esperado:**
```text
True
False
```

## reversed()

**Descripción y uso:** Devuelve un iterador que recorre una secuencia en orden inverso, sin modificarla.

**Sintaxis:**
```python
reversed(secuencia)
```

**Parámetros:** secuencia: objeto reversible, como una lista o una tupla.

**Ejemplo práctico:**

````python
letras = ["a", "b", "c"]
print(list(reversed(letras)))
print(letras)
````

**Resultado esperado:**
```text
['c', 'b', 'a']
['a', 'b', 'c']
```

## format()

**Descripción y uso:** Formatea un valor como cadena según una especificación. También se usa con marcadores de formato.

**Sintaxis:**
```python
format(valor, especificacion='')
```

**Parámetros:** valor: dato que se formatea; especificacion: formato opcional, por ejemplo cantidad de decimales.

**Ejemplo práctico:**

````python
pi = 3.141592
print(format(pi, ".2f"))
print("El total es {}".format(25000))
````

**Resultado esperado:**
```text
3.14
El total es 25000
```

## isinstance()

**Descripción y uso:** Comprueba si un objeto pertenece a una clase o a una tupla de clases. Devuelve True o False.

**Sintaxis:**
```python
isinstance(objeto, clase)
```

**Parámetros:** objeto: valor que se evalúa; clase: tipo o tupla de tipos permitidos.

**Ejemplo práctico:**

````python
dato = 42
print(isinstance(dato, int))
print(isinstance(dato, (int, float)))
print(isinstance(dato, str))
````

**Resultado esperado:**
```text
True
True
False
```

**Sintaxis:**
```python

```

**Parámetros:** objeto: elemento sobre el que se desea consultar la documentación; si se omite, abre la ayuda interactiva.
